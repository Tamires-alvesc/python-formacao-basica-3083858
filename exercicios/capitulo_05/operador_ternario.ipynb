- Reescreva a estrutura condicional do exercício do arquivo "if_elif_else.ipynb" usando operador ternário


```python
from random import randint

lista30 = []

for c in range(0,30):
    n = randint(0,100)
    lista30.append(n)
print(lista30)


ate_12 = [] 
entre_13_e_20 = []
maior_21 = []


for item in lista30:
    ate_12.append(item) if item < 12 else entre_13_e_20.append(item) if 13 <= item < 20 else maior_21.append(item) if item >= 21 else print('Não tem uma lista')

print(ate_12)
print(entre_13_e_20)
print(maior_21)
```
