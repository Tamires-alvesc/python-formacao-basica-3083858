- Crie uma lista


```python
lista = ['a', 'b', 'c', 'd', 'e', 'f']

print(lista)
```

- Crie um for loop para imprimir na tela todos os elementos da lista


```python
for x in range(0,len(lista)):
    print(lista[int(x)])

for item in lista:
    print(item)
```

- Crie um for loop usando o recurso "continue"


```python
lista1 = [1, 2, 3, 4, 5, 6, 7]

for item in lista1:
    if item <= 4:
        continue

    print(item)
```

- Crie um for loop usando o recurso "break"


```python
lista1 = [1, 2, 3, 4, 5, 6, 7]

for item in lista1:
    print(item)
    if item >= 4:
        break

  
```

- Crie um for loop usando o "else"


```python
lista1 = [1, 2, 3, 4, 5, 6, 7]

for item in lista1:
    if item <= 4:
        continue
    print(item)
else:
    print('Agora estou aqui')
```
