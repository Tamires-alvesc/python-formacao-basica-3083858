- Crie uma classe com 3 atributos


```python
class Roupa:
    def __init__(self, cor ='branco', tipo='camisa', genero='unissex'):
        self.cor = cor
        self.tipo = tipo
        self.genero = genero
```

- Defina uma variável de classe


```python
roupa_festa = Roupa(rosa, vestido, feminino)
```

- Modifique a variável de instância "cor"


```python
roupa_festa.cor = 'azul'
```

- Printe na tela a variável de classe "cor"


```python
print(roupa_festa.cor)
```

```python
class Roupa:
    def __init__(self, cor ='branco', tipo='camisa', genero='unissex', preco):
        self.cor = cor
        self.tipo = tipo
        self.genero = genero
        self.__preco = preco
```
