- Crie uma função que receba como argumento 2 números e 1 operação matemática


```python
def operacao_matematica(a=0, b=0, op='soma'):
    if op == 'soma':
        response = a + b
    elif op == 'multiplicação':
        response = a*b

    return response

print(operacao_matematica(a=3,b=4))
print(operacao_matematica(3,4,'multiplicação'))
```

- Modifique a função para que receba uma quantidade de argumentos não previsível e os imprima na tela


```python
def operacao_matematica(*nums, op):
    if op == 'soma':
        response = sum(nums)
    elif op == 'multiplicação':
        multi = 1
        for n in nums:
            multi *= n
        response = multi

    return response
    
print(operacao_matematica(3,4,5,6,7, op ='soma'))
print(operacao_matematica(2,3,4, op ='multiplicação'))
```

- Use a biblioteca math para otimizar a função


```python
import math

def operacao_matematica(*nums, op):
    if op == 'soma':
        response = sum(nums)
    elif op == 'multiplicação':
        response = math.prod(nums)

    return response
    
print(operacao_matematica(3,4,5,6,7, op ='soma'))
print(operacao_matematica(2,3,4, op ='multiplicação'))
```
