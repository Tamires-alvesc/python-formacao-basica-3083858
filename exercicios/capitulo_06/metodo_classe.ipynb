- Crie uma classe


```python
class Moradia:
    def __init__(self, tipo, tamanho, localidade, valor, nivel):
        self.tipo = tipo
        self.tamanho = tamanho
        self.localidade = localidade
        self._valor = valor
        self._nivel = nivel

    def getValor(self):
        return self._valor

    def setNivel(self, novo_nivel = "milionario"):
        if novo_nivel:
            self._nivel = novo_nivel
        return self._nivel
```

- Estabeleça variáveis privadas no construtor 


- Crie um método get, que retorne uma das variáveis privadas


- Crie um método set, que modifique uma das variáveis privadas


- Instancie a classe e imprima na tela a variável antes e depois da modificação


```python
lar = Moradia('apartamento', 'grande', 'capital', '1milhao', 'luxo')
print(lar._nivel)
print(lar.setNivel())
```
