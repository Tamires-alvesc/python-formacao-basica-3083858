- Converta uma lista em um conjunto


```python
lista1 = [1, 2, 2, 3, 4, 4, 5, 'banana', 'Tamires', 'banana']
novo_conjunto = set(lista1)
print(novo_conjunto)
```

- Insira novos valores no conjunto


```python
novo_conjunto.add(12)
novo_conjunto.update([4,7])

print(novo_conjunto)
```

- Remova um elemento do conjunto


```python
novo_conjunto.remove(12)
print(novo_conjunto)
```

- Verifique se um elemento está contino no conjunto

```python
item = 9
resposta = f'{item} não está no conjunto' 
if item in novo_conjunto:
    resposta = f'{item} está no conjunto'
print(resposta)

    
```
